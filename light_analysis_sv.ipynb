```python
import pandas as pd
import plotly.express as px
import math

# Load all datasets
datasets = {
    "single_128": pd.read_csv('/scratch2/mrenaudin/colorlessgreenRNNs/val_loss/cbr1h128_shuffling/epoch40'),
    "single_256": pd.read_csv('/scratch2/mrenaudin/colorlessgreenRNNs/val_loss/cbr1h256_shuffling/epoch40'),
    "single_512": pd.read_csv('/scratch2/mrenaudin/colorlessgreenRNNs/val_loss/cbr1h512_shuffling/epoch40'),
    "two_128": pd.read_csv('/scratch2/mrenaudin/colorlessgreenRNNs/val_loss/cbr2h128_shuffling/epoch40'),
    "two_256": pd.read_csv('/scratch2/mrenaudin/colorlessgreenRNNs/val_loss/cbr2h256_shuffling/epoch40'),
    "two_512": pd.read_csv('/scratch2/mrenaudin/colorlessgreenRNNs/val_loss/cbr2h512_shuffling/epoch40'),
    "eight_128": pd.read_csv('/scratch2/mrenaudin/colorlessgreenRNNs/val_loss/cbr8h128_shuffling/epoch40'),
    "eight_256": pd.read_csv('/scratch2/mrenaudin/colorlessgreenRNNs/val_loss/cbr8h256_shuffling/epoch40'),
    "eight_512": pd.read_csv('/scratch2/mrenaudin/colorlessgreenRNNs/val_loss/cbr8h512_shuffling/epoch40'),
}

# Concatenate all with appropriate metadata
all_dfs = []
for name, df in datasets.items():
    depth, size = name.split("_")
    df = df.copy()
    df["depth"] = depth  # 'single', 'two', 'eight'
    df["size"] = size    # '128', '256', '512'
    df['val_loss']= np.exp(df['val_loss'])
    all_dfs.append(df)

df_all = pd.concat(all_dfs)

# Plot with color by depth, line dash by hidden size
fig = px.line(
    df_all, 
    x="epoch", 
    y="val_loss", 
    color="depth", 
    line_dash="size",
    title="Validation Perplexity by Number of Heads and Hidden Size for CBR RNN"
)

fig.update_layout(
    xaxis_title="Epoch",
    yaxis_title="Perplexity",
    template="plotly_white",
    legend_title="Architecture"
)

fig.show()
```
