- Does performance on long range agreement and retention over filled interval correlate ? 
- Hyp : If the model performs poorly on retention over filled interval, then it won't remember the subject when processing the verb, and will be more sensitive to interference from the attractor. 
- Test : Correlation between performance on incongruent cases of NounPP and retention over filled interval

```python
import torch
from scipy.stats import spearmanr
```

```python
nounpp = torch.load('/scratch2/mrenaudin/colorlessgreenRNNs/evaluation_notebooks/results/test_attention_650')
rfl = torch.load('/scratch2/mrenaudin/colorlessgreenRNNs/wm_tests/results/test_attention_650_sce3')
```

```python
median_repeat= {}
for i in rfl.keys():
    median_repeat[i]= rfl[i]['repeat']['list len : 7, prompt len : 1'].median()
```

```python
epochs = sorted(median_repeat.keys())

# Convert tensors to floats
repeat_vals = [median_repeat[e].item() for e in epochs]

# Compute Spearman for each condition
for condition in nounpp[epochs[0]]:
    condition_vals = [nounpp[e][condition] for e in epochs]
    corr, pval = spearmanr(repeat_vals, condition_vals)
    print(f"Spearman correlation between median_repeat and '{condition}': {corr:.3f} (p={pval:.4f})")
```

 So when repeat surprisal increases, performance on incongruent cases decreases ! Test passed ! P values are not too bad for incongruent cases, and correlation are about the strongest also
