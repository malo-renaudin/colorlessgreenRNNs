```python
import torch.nn as nn
```

```python
class Transformer(nn.Module):
    def __init__(self, ntoken, ninp, nhid, nheads, ndecoder_layers, d_feedforward, dropout=0.5, device=None):
        super().__init__()
        self.ndecoder_layers = ndecoder_layers
        self.device = device
        self.nheads = nheads
        self.dropout = dropout
        self.encoder = nn.Embedding(ntoken, ninp)
        self.Transformer = nn.Transformer(nhid, nheads, num_encoder_layers=0, num_decoder_layers=ndecoder_layers, dropout=dropout, dim_feedforward=d_feedforward)
```
